## 2. Download Apple stock data

We request daily prices for **AAPL** (Apple Inc.) from **January 1, 2020** through **today**.

The returned DataFrame includes: `Date`, `Open`, `High`, `Low`, `Close`, and `Volume`.

```python
ticker = "AAPL"
start_date = "2020-01-01"
end_date = date.today().strftime("%Y-%m-%d")

df = get_stock_data(ticker, start_date, end_date)

print(f"Downloaded {len(df)} rows of {ticker} data ({start_date} to {end_date})")
```
